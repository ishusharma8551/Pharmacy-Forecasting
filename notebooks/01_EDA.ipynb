## Nine charts + insights (insights are computed from the data, not hard-coded)

```
meds = pd.read_csv(config.PROCESSED_DIR/'medicines_clean.csv')
batches = pd.read_csv(config.PROCESSED_DIR/'batches_clean.csv', parse_dates=['Purchase_Date','Manufacturing_Date','Expiry_Date'])
sales = pd.read_csv(config.PROCESSED_DIR/'sales_daily_clean.csv', parse_dates=['Date'])
insights = eda.run_eda(sales, meds, batches, sales['Date'].max())
```

```
for i, (name, text) in enumerate(insights.items(), 1):
    f = sorted((config.REPORTS_DIR/'figures').glob(f'{i:02d}_*.png'))[0]
    display(Markdown(f'### {name}\n{text}')); display(Image(filename=str(f)))
```
